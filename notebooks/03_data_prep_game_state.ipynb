```python
import pandas as pd
import numpy as np
```

```python
df_raw = pd.read_csv("data/raw/raw_game_state_data.csv")
df = df_raw.copy()
```

```python
df
```

```python
df["goals_minus_xG"] = df["goals"] - df["xG"]
df["goals_against_minus_xGA"] = df["goals_against"] - df["xGA"]
```

```python
df.head()
```

```python
df["minutes"] / df["goals_against"]
```

```python
df["total_goals"] = df.groupby(["season","team"])["goals"].transform("sum")
df["total_goals_against"] = df.groupby(["season", "team"])["goals_against"].transform("sum")
```

```python
df["total_game_time"] = df.groupby(["season", "team"])["minutes"].transform("sum")
```

```python
df["game_state_percentage"] = df["minutes"] / df["total_game_time"] *100
```

```python
df["goals_percentage"] = df["goals"] / df["total_goals"] *100
df["goals_against_percentage"] = df["goals_against"] / df["total_goals_against"] *100
```

```python
df["total_xG"] = df.groupby(["season", "team"])["xG"].transform("sum")
df["total_xGA"] = df.groupby(["season", "team"])["xGA"].transform("sum")

df["xG_percentage"] = df["xG"] / df["total_xG"] *100
df["xGA_percentage"] = df["xGA"] / df["total_xGA"] * 100
```

```python
df
```

```python
mapping = {
    "Goal diff > +1": "Winning (2+ goals)",
    "Goal diff +1": "Winning (1 goal)",
    "Goal diff 0": "Drawing",
    "Goal diff -1": "Losing (1 goal)",
    "Goal diff < -1": "Losing (2+ goals)"
}
```

```python
df["game_state"] = df["game_state"].replace(mapping)
```

```python
df
```

```python
cols_to_keep = ["season", "team", "game_state", "minutes", "total_game_time", "game_state_percentage",
 "goals", "total_goals", "goals_percentage", "xG", "total_xG", "xG_percentage",
 "goals_against", "total_goals_against", "goals_against_percentage", "xGA", "total_xGA", "xGA_percentage"]
```

```python
df[cols_to_keep].to_csv("data/processed/cleaned_game_state_data.csv", index=False)
```
