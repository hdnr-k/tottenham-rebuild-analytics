```python
import pandas as pd
import numpy as np
```

```python
df_raw = pd.read_csv("data/raw/raw_match_shots_data.csv")
df = df_raw.copy()
```

```python
df.info()
```

```python
df = df.sort_values(by=["date", "minute"]).reset_index(drop=True)
```

```python
df.tail()
```

```python
# トッテナムがアウェイ（a_team)の試合で、ホームの(h)のチームが打ったシュート
cond_away_match = (df["a_team"] == "Tottenham") & (df["h_a"] == "h")
# トッテナムがホーム（h_team)の試合で、アウェイ(a)のチームが打ったシュート
cond_home_match = (df["h_team"] == "Tottenham") & (df["h_a"] == "a")

df_conceded_shots = df[cond_home_match | cond_away_match].reset_index(drop=True)
```

```python
df_conceded_shots.to_csv("data/processed/cleaned_match_shots_data.csv", index=False)
```
